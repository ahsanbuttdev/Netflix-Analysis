```python
import pandas as pd

df1 = pd.read_csv(r"D:\datasets\PortfolioProjects-main\netflix_dataset\netflix_tv_shows_detailed_up_to_2025.csv")

df2 = pd.read_csv(r"D:\datasets\PortfolioProjects-main\netflix_dataset\netflix_movies_detailed_up_to_2025.csv")
```

```python
main_df = pd.concat([df1, df2], ignore_index = True)

main_df['show_id'] = main_df['show_id'].astype(str)

main_df['show_id'] = main_df.apply(lambda row: f"M-{row['show_id']}" if row['type'] == 'Movie' else f"T-{row['show_id']}", axis=1)

main_df
```

```python
main_df.info()
```

```python
bridge_df = main_df[['show_id', 'country']].copy()

bridge_df['country'] = bridge_df['country'].fillna('Unknown') 

bridge_df['country'] = bridge_df['country'].str.split(',')

bridge_df = bridge_df.explode('country')

bridge_df['country'] = bridge_df['country'].str.strip()

bridge_df
```

```python
main_df = main_df.drop(columns = 'country')

main_df
```

```python
cast_df = main_df[['show_id', 'cast']].copy()

cast_df['cast'] = cast_df['cast'].fillna('Unknown') 

cast_df['cast'] = cast_df['cast'].str.split(',')

cast_df = cast_df.explode('cast')

cast_df['cast'] = cast_df['cast'].str.strip()


main_df = main_df.drop(columns = 'cast')

main_df
```

```python
director_df = main_df[['show_id', 'director']].copy()

director_df['director'] = director_df['director'].fillna('Unknown') 

director_df['director'] = director_df['director'].str.split(',')

director_df = director_df.explode('director')

director_df['director'] = director_df['director'].str.strip()

main_df = main_df.drop(columns = 'director')

main_df
```

```python
genres_df = main_df[['show_id', 'genres']].copy()

genres_df['genres'] = genres_df['genres'].fillna('Unknown') 

genres_df['genres'] = genres_df['genres'].str.split(',')

genres_df = genres_df.explode('genres')

genres_df['genres'] = genres_df['genres'].str.strip()

main_df = main_df.drop(columns = 'genres')

genres_df
```

```python
main_df['description'] = main_df['description'].fillna('Description not available')

main_df['duration'] = main_df['duration'].fillna('Unknown')
main_df
```
